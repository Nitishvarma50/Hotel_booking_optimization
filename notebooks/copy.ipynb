```python
df['nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
df['guests'] = df['adults'] + df['children'].fillna(0) + df['babies']
```

```python
def show(data, title):
    city = data[data['hotel'] == 'City Hotel']['is_canceled'].mean()*100 
    resort = data[data['hotel'] == 'Resort Hotel']['is_canceled'].mean()*100
    both = data['is_canceled'].mean()*100
    print(title,"\n", len(data),"\n", city,"\n", resort,"\n", both)

show(df, "0. raw")
```

##### Removing 40k rows as this are duplicates

```python
df = df.drop_duplicates()   # 1. remove exact duplicates

show(df, "1. no duplicates")
```

```python
df[df['adr'] < 0].shape
```

##### Removing 1 row as the adr cannont be negative

```python
df = df[df['adr'] >= 0]              # 2. remove negative price
show(df, "2. no negative adr")
```

### Removing ADR Greater Then 1000 as they are outliers and their value is very High

```python
df = df[df['adr'] <= 1000]
```

```python
df = df[df['adr'] <= 1000] # 3. remove the 5400 price
show(df, "3. no adr above 1000")
```

```python
df['guests'] = df['adults'] + df['children'] + df['babies']
zero_guest_rows = df[df['guests'] == 0]
print(len(zero_guest_rows))  
```

```python
df = df[df['guests'] > 0]            # 4. remove zero guests
show(df, "4. no zero guests")
```

```python
df = df[df['nights'] > 0]            # 5. remove zero nights
show(df, "5. no zero nights")
```

Filling Null values

```python
print("\nSTEP 1: missing values before fixing")
missing = df.isna().sum()
print(missing[missing > 0])
```

```python
print("\nSTEP 2: children")
print("  blanks before:", df['children'].isna().sum())
df['children'] = df['children'].fillna(0).astype(int)
print("  blanks after: ", df['children'].isna().sum())
```
